```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(1)
```

## Task 1: a distribution

Each step: write the prediction on the sheet, run the cell, compare.

```python
# 1. Ten thousand numbers. Predict the mean and the standard deviation before you run.
x = rng.normal(100, 15, 10000)
print(x.mean(), x.std())
```

```python
# 2. The histogram. Predict the shape, the peak and the width before you run.
fig, ax = plt.subplots()
ax.hist(x, bins=40)
ax.set_xlabel("value")
ax.set_ylabel("count")
fig.savefig("normal.png")   # a notebook runs in its own folder, so no module-4/ in the path; the figure also appears below the cell
```

```python
# 3. Mark the mean and one standard deviation either side. Predict the fraction between the dashed lines, then check it with the mask.
fig, ax = plt.subplots()
ax.hist(x, bins=40)
ax.axvline(x.mean(), color="k")
ax.axvline(x.mean() + x.std(), color="k", linestyle="--")
ax.axvline(x.mean() - x.std(), color="k", linestyle="--")
ax.set_xlabel("value")
ax.set_ylabel("count")
fig.savefig("normal_marked.png")
print((abs(x - x.mean()) < x.std()).mean())
```
