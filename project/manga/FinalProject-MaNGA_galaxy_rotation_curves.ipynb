## What is a rotation curve?

A [rotation curve](https://en.wikipedia.org/wiki/Galaxy_rotation_curve#:~:text=The%20rotation%20curve%20of%20a,distance%20from%20that%20galaxy's%20centre.) measures the speed of stars or gas orbiting inside a galaxy as a function of the distance from the galaxy's center.  While each star or gas cloud in a galaxy may be on its own independent orbit, in a _rotating_ galaxy, all of the orbits tend to all go in the same direction.  The result is a disk-like galaxy that appears to be "rotating" even though it is not a solid object and material at different radii rotate at different speeds.

Galaxies are held together by gravity.  Imagine a star (like our sun) orbiting at some given radius in a galaxy.  What keeps the star from just flying off into intergalactic space?  Well, kind of analogous to a [tetherball](https://en.wikipedia.org/wiki/Tetherball), the star feels an inward pull from the gravity of all the galaxy's mass interior to the star's location.  

We can think of the star as having "settled" at an orbital radius that balances the star's kinetic energy of motion against the potential energy it feels from all the gravitating mass interior to that radius.  In physics terms:

$$ m_{\rm star} v^2 = \frac{G M_{\rm interior} m_{\rm star}}{r} $$

You can see how $m_{\rm star}$, the mass of the star, cancels out on both sides.  Then, if we knew $r$ (distance from the galaxy's center) and $v$ (the star's orbital velocity), we could solve for $M_{\rm interior}$, the mass of the galaxy interior to that location.

Well, a plot of $v$ versus $r$ is basically a rotation curve! So, you can see now how a rotation curve lets us "weigh" the mass of a galaxy.

What we'll find out is that rotation curves reveal that there's a lot more mass than just the stars and gas that we observe in galaxies.  There must be something else that doesn't shine and doesn't obscure... something like... dark matter!

## Why is MaNGA so good for measuring rotation curves?

For a sample of 10,000 galaxies, the MaNGA survey obtained a spectrum at each of 100-1000 points across the face of each galaxy.  This allowed us to construct "data cubes" with dimensions $(x, y, \lambda)$ for each galaxy.  The data cube tells us the measured flux at each $(x,y)$ position across the face of the galaxy as a function of wavelength, $\lambda$.

Atoms, like Hydrogen, emit light at specific, well-known wavelengths.  But if those emitting atoms are moving towards or away from us, the doppler shift induces a velocity-dependent offset in the observed wavelength compared to the intrinsic one.  It is easy to measure this shift in a spectrum.  

Therefore, a spectrum at every position in a galaxy (like MaNGA provides) can tell us the line-of-sight velocity of Hydrogen (and other galaxy components) at every position.  

So, with the 2D spectral maps that MaNGA provides, you can see we are in good shape for measuring rotation curves from MaNGA galaxies.
